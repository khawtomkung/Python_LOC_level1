```python
"for loop"
# 1. loop by index - เกี่ยวกับตัวเลข
nf = "Namfon"
for i in range(len(nf)): # range 0,1,2,3,4,5 range len(nf)
    print(nf[i])

# 2. loop by element - เกี่ยวกับตัวอักษรอย่างเดียว
for character in nf: # character = ตัวแปร
    print(character)

# 3. loop by enumerate(both index,element)
for i, ele in enumerate(nf): # index(i) element(ele) คู่กัน
    print(i, ele)
```

```python
# จงหาตัวอักษรที่แตกต่างกันทั้งหมดในสตริง s

s = "namfon"
set(s)
len(set(s))

found = set() # เอาไว้เก็บตัวที่เคยเจอมาแล้ว
result = 0
for character in s:
    if character not in found:
        result += 1
        found.add(character)
result
```
