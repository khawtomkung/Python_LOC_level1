```python
import time
n = 40
def fibo(n):
    if n == 1 or n == 2:
        return 1
    return fibo(n-1) + fibo(n-2)

t1 = time.time()
fibo(n)
t2 = time.time()
t2 - t1
```

```python
t1 = time.time()
n = 8
dp = [0 for _ in range(n)] # ประกาศ list มีแต่เลข 0 จำนวน n ครั้ง
dp[0] = 1
dp[1] = 1

for i in range(n):
    if i > 1:
        dp[i] = dp[i-1] + dp[i-2]

dp[n-1]
t2 = time.time()
t2 - t1
```
