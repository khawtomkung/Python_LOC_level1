```python
# try except(ลองทำเจอ error แล้วทำอะไรต่อ)

try:
    a = int(input())
    b = 10 / a
except Exception as e: # ครอบคลุมทุก Error(ทุกข้อผิดพลาด)
    print(e) # e เป็นตัวแปร = division by zero
```

```python
try: # ทำแน่ๆ
    a = 0
    b = 0 / 0
except Exception: # ถ้ามี Error จะทำอันนี้
    print("Noooo I hate 0")
finally: # ไม่ว่ายังไงก็จะทำ ทุกกรณี
    print("yayyy")
```

```python
# assert (เช็คค่าว่าโปรแกรมทำงานถูกต้องไหม, เอาไว้แก้โค้ด)

a = 0
assert a == 1, "try again" # AssertionError: try again
print(a)
```

```python
# a = 50
class CustomErr(Exception):
    def _init_(self, message):
        super()._init_(message)

a = int(input())
b = 10 - a
if b < 0:
    raise CustomErr("Bro")
else:
    print(b)

# raise CustomErr("bro") # ถ้ามีกรณีนี้เกิดขึ้นให้ทำสิ่งนี้เลย
```

```python
# Add 10 to any number

"""
a = 10
b = a + 10
c = b + 10
d = c + 10
print(a,b,c,d)
"""

def AddTen(num1):
    print("hi", end=" ")
    print(num1 + 10)
    return num1 + 10

a = 10
b = AddTen(a)
c = AddTen(b)
d = AddTen(c)
```

```python
# function(คำสั่ง หรือชุดคำสั่ง เรียกใช้ได้เรื่อยๆที่เรียก), argument(ใส่), parameter(รับ)

# 1. No input, No return
def f0():
    print("hi")
f0()

# 2. input, no return
def f1(para): # para is a 'parameter' of f1 function
    print(para)
f1("hello") # "hi" is an 'argument' of f1 function แทนที่ 'parameter'

# 3. no input, return
def f2():
    return 2
a = f2() # a = 2

# 4. input, return (เต็ม)
def f3(num):
    return num + 2
b = f3(a)
b # a + 2 = 4
```
