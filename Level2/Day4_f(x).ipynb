```python
# function

def name_of_the_function(input): # input = parameter
    return input # return = output
name_of_the_function("argument") # "argument" = argument ที่แทนที่ใน parameter
```

input -> 0, 1, many inputs
output -> 0, 1, many outputs

```python
# 1. input 1, output 1
# square -> f(x) = x ** 2
def square(a):
    print('กำลังคำนวณยกกำลังสอง')
    result = a ** 2
    return result # output result

(square(7)) # 7 ** 2 = 49

# 2. input 1, output 0(None)
def greeting(name): # name = parameter
    print("hello " + name)
    return # None or no return

greeting('Rimuru') 

# 3. input 1, output many
def get_numbers(n):
    return n, n+1, 2*n+2

get_numbers(3) # (2, 3, 6) = tuple = ()

# 4. input 0, output 0
def greeting3():
    print('Hello No Name!')

def check_current_version():
    import time
    return time._version_()

# 5. input many
def add(a, b): # a, b = parameter
    return a + b

add(1,2)

def multiply(a, b, c, d):
    result = a*b*c*d
    return result

multiply(1,2,3,4)
```

```python
# position argument, keyword argument("**ห้ามนำหน้า positional argument เสมอ**")
def add(a, b=2): # b=2 is keyword argument
    return a + b

add(3)
```

```python
def multiply(a, b, c=2, d=3, e=4):
    return a*b*c*d*e

multiply(2,3,4,5,6) # position argument = argument เชิงตำแหน่ง หน้าไปหลัง
# 2*3*4*5*6
multiply(2,3,4) # 2*3*4*3*4
multiply(2,4) # 2*4*2*3*4
multiply(2,4,e=5) # 2*4*2*3*5
multiply(1,3,6,e=3) # a=1 b=3 c=6 d=3 e=3
multiply(1,2,c=4,d=5) # a=1 b=2 c=4 d=5 e=4
```
