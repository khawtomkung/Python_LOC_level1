*args(non-keyword argument) and *kwargs(keyword argument)

```python
# *args tuple - รับกี่ค่าก็ได้ ไม่มีชื่อ
def fun(*a):
    print(a) # ('1', '2', 'มังคุด')
    for arg in a:
        print(arg)

fun('1','2','มังคุด')
```

```python
# **kwargs dictionary(key, value) - รับมาเป็นคู่
def fun(**a):
    print(a) # {'one': '1', 'two': '2', 'three': '3'}
    for key, value in a.items():
        print(key, value)

fun(one='1',two='2',three='3')
```

```python
# Literal decaration - กำหนดค่าที่รับมา
def func(a : str) -> str: #รับค่ามาเป็น str ส่งค่าออกเป็น str
    a.count #libary str (method)
    print(a.count)
    return a.count
newstr = func("hello")
newstr
```

gcd() - หารร่วมมาก
lcm() - คูณร่วมน้อย
gcd(a,b) * lcm(a,b) = a * b

```python
# gcd()
# 1. if(ช้า)
import time

a = 6 #min
b = 62 #max
def gcd1(a,b):
    for i in range(min(a,b),1,-1): #หาตัวที่มากสุด เลยเริ่มจากตัวที่มากที่สุด
        if a % i == 0 and b % i == 0: #if not(a % i) and not(b % i) :
            print(i)
            break
"""gcd1(12, 15)"""

# 2. euclid
a = 12
b = 10
def gcd2(a,b): # รับค่า ไม่ส่งค่า
    while 1:
        r = a % b
        if r == 0:
            break
        a = b
        b = r
    return b

gcd = gcd2(a,b)
lcm = a * b / gcd
print(gcd, int(lcm))
```
