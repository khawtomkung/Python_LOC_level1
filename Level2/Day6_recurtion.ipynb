```python
# recursive function "worst algorithm in term of time complexity(*Depth first search*)" นาน 
def fibo(n): # เรียกตัวเองซ้ำ
    #base case ต้องมีเสมอ ไม่งั้นจะ loop ไม่สิ้นสุด
    if n == 1 or n == 2:
        return 1
    #recurtion
    return fibo(n-1) + fibo(n-2)
ans = fibo(40) # หาตำแหน่งที่ n
ans
```

factorial(n) = n * n(n-1) * ... * 1 | 5! = (5 * 4 * 3 * 2 * 1) = 120 0! = 1

```python
def factorial(n):
    #base case***
    if n == 0:
        return 1
    #recurtion
    return n * factorial(n-1)
factorial(5)
```

```python
# gcd lcm
def gcd(a,b):
    #base case
    if b == 0:
        return a
    #recurtion
    return gcd(b,a%b) #euclid
ans = gcd(132,216)
ans

# this requires gcd, *args, recursive knowledge
def gcd(*args): #args input infinity to tuple "()"
    #base case
    if len(args) == 1: #ตัวเดียว ออกมาเป็นตัวนั้น
        return args[0]
    elif len(args) == 2: #2 ตัว
        a, b = args
        if b == 0:
            return a
        else:
            return gcd(b,a%b)
    #recurtion
    else: #tuple a *rest ... 3 ตัวขึ้นไป
        first, second, *rest = args
        return gcd(gcd(first,second),*rest)
    
ans = gcd(120,234,28,34) #รับค่าตัวเดียว = ค่านั้น
ans

# gcd(120,234) => x
# gcd(x,28) => y
# gcd(y,34)
```
