```python
# basic usecase example of class
class Real_Student:
    def __init__(self, age, height, weight): #self property
        self.age = age
        self.height = height
        self.weight = weight
        self.species = "homosapien"

    def BMI(self):
        self.bmi = self.weight / (self.height ** 2)
        return self.bmi

K = Real_Student(17, 1.61, 45)
C = Real_Student(16, 1.52, 40)

print(K.BMI(), C.BMI(), C.age, K.species)
```

```python
class Triangle:
    def __init__(self, base, height):
        self.base_length = base
        self.height_length = height
    def area_cal(self):
        return 0.5 * self.base_length * self.height_length
    def multiply_area(self, multi): #ใส่ค่าที่เอาไปคูณ
        return 0.5 * self.base_length * self.height_length * multi

Tri1 = Triangle(5,4)
print(Tri1.area_cal())

print(Tri1.multiply_area(3))
```
