Class: 
A blueprint for creating objects. Defines a set of attributes and methods that the created objects can use.

Object: 
An instance of a class. When a class is defined, no memory is allocated until an object of that class is created.

Attribute:
A variable that belongs to an object or class. It is used to store data relevant to the object.

Method:
A function that belongs to an object or class. It is used to perform operations on the data within the object.

self:
A reference to the current instance of the class. It is used to access variables and methods associated with the object.

```python
a = 1
isinstance(a,int) # อยู่ใน class int ใช่มั้ย
isinstance(a,str) # อยู่ใน class str ใช่มั้ย
```

```python
class Animal: #Class
    def __init__(self, name, species): #__init__ เก็บข้อมูล(Attribute)
        self.name = name  # Attribute
        self.species = species  # Attribute

    def speak(self,speech):  # Method
        print(f"My name is {self.name} and I am a {self.species}, {speech}")
        
dog = Animal("hottodogu", "Dog") #Object
cat = Animal("maew sommm", "Cat")

dog.speak('hi')
cat.speak('meowwwwwwwww')
print(isinstance(dog,Animal)) # dog อยู่ใน class Animal มั้ย
print(dog.name)
```

```python
#Add more Atribute and Method
class Animal:
    def __init__(self, name, species, sound):
        self.name = name
        self.species = species
        self.sound = sound

    def speak(self):
        print(f"My name is {self.name} and I am a {self.species} and I say {self.sound}")
    
    def rename(self, new_name):
        self.name = new_name

dog = Animal("hottodogu", "Dog","woof woofff")
cat = Animal("maew sommm", "Cat","meowwwwwww")

dog.speak()  
cat.speak()

dog.rename("muffinnn")
dog.speak()
```

## Inheritance(การสืบทอด)

```python
class Animal: #Base Class
    def __init__(self, name, species):
        self.name = name #Attribute
        self.species = species

    def speak(self):
        print(f"My name is {self.name} and I am a {self.species}")

# Dog class inherits from Animal class  // Sub-Class

#every dog is an animal
#not every animal is a dog

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "Dog")
        #self.name
        #self.species
        self.breed = breed #Attribute of class Dog

    def speak(self):
        print(f"My name is {self.name} and I am a {self.breed} dog")

# Creating an object of Dog class
dog = Dog("Buddy", "Golden Retriever")
dog.speak()  # Output: My name is Buddy and I am a Golden Retriever dog
print(dog.name)
```

## Additional
- Encapsulation(ห้ามแก้ไขนอก class / Private Attribute)
- Polymorphism(การใช้ method ซ้ำกันใน sub class)
- Class and Static Methods()

```python
#Encapsulation -> ห้ามแก้ไข Attribute บางตัว
class Animal:
    def __init__(self, name, species):
        self.__name = name #Private Attribute
        self.species = species

    def speak(self):
        print(f"My name is {self.__name} and I am a {self.species}")

    def get_name(self): #getter method of name
        return self.__name
    
    def set_name(self,name): #setter method of name
        self.__name = name
        
bird = Animal("tweety","bird")
bird.speak()
print(bird.get_name())
bird.set_name("tweetooo")
print(bird.get_name())
```

```python
#Polymorphism -> method ชื่อซ้ำใน subclass
class Animal:
    def __init__(self, name):
        self.name = name
        
class Dog(Animal):
    def speak(self):
        return f"hi im {self.name}, woof!"
    

class Cat(Animal):
    def speak(self):
        return f"hi im {self.name}, Meowwwwwwwww!"
    
dog = Dog("muffinn")
cat = Cat("Meawjomm")

for ani in (dog, cat): #polymorphism
    print(ani.speak())
```

```python
#Class Method and Static Method

class Animal:
    kingdom = "Animalia"
    
    def __init__(self, name):
        self.name = name
        
    def get_kingdom(cls): #class method
        return cls.kingdom
    
    def action(a,act): #Static metod
        print(f"I'm currently {act} right now")
    
cat = Animal("khaw")
print(cat.get_kingdom())
cat.action('jumping')
    
```
