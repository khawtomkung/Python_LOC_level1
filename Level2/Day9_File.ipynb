Different modes you can use with the open() function:
'r' : Read mode (default)
'w' : Write mode
'a' : Append mode
'r+' : Read and write mode
'w+' : Write and read mode (truncate the file)
'a+' : Appent and read mode
'rb' : Read mode in binary
'wb' : Write mode in binary
'ab' : Append mode in binary
'rb+' : Read and write mode in binary
'wb+' : Write and read mode in binary
'ab+' : Append and read mode in binary

```python
# Example for using file
with open('./file/new.txt', 'r') as file: #Read
    lines = file.readlines() #()ปิดไฟล์ให้

modified_lines = [line.upper() for line in lines] #Modify : make each line

with open('./file/modified_new.txt', 'w') as file: #Write
    file.writelines(modified_lines)
```

# seek(line_index)
""" กลับไปอ่านบรรทัดที่ line_index + 1 """

```python
with open('./file/new.txt', 'r') as file:
    file_read = file.read()
    file.seek(0) # กลับไปอ่านใหม่
    file_readline = file.readline()
    file_readlines = file.readlines()
print(file_read)
print("......")
print(file_readline)
print("......")
print(file_readlines)
```

# write()
""" เขียนทับ, แทนที่ไฟลทั้งไฟล์ """

```python
with open('./file/modified_new.txt', 'a') as file:
    file.write('hello')
```

# append()
""" เพิ่มเข้าไปในไฟล์ """

```python
with open('./file/modified_new.txt', 'w') as file:
    file.write('hello')
```

# write/append without write()

```python
with open('./file/modified_new.txt', 'r+') as file:
    content = file.read()
    content += "Hello\n"
    content += "This is\n"
    content += "New content\n"
    file.write(content)

with open('./file/modified_new.txt', 'w')  as file:
    file.write(content)
```
