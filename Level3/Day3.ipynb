```python
import numpy as np

arr1 = np.array([3,4,5])
arr2 = np.array([7,8,9])
```

## Logical Operations

```python
# 1 Elemwent Equality
arr1 == arr2 # array([False, False, False])

np.all(arr1==arr2) # True
np.all(arr1>=arr2) # False
# Multiple Equalities
np.all(arr1<6) # True
# all of the elements less than x
np.all(arr1<5) # False
# any of the elements less than x
np.any(arr1<4) # True
# any of elements is evn number
np.any(arr1 % 2 == 0)
```

## Changing Elements

```python
l = [1, 1, 1, 1]
l[2] = 2
l

arr3 = np.array([1, 1, 1, 1])
arr3[2] = 2
arr3 # array([1, 1, 2, 1])

arr3 = np.array([[1, 2, 3], [4, 5, 6]])
arr3, arr3.shape # (2, 3)
arr3[0][2] = 7
arr3 # array([[1, 2, 7], [4, 5, 6]])
```

## Matrix Multiplication

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]]) # (2, 3)
arr2 = np.array([[-1, -2], [-3, -4], [-5, -6]]) # (3, 2)

np.matmul(arr1, arr2) # (2, 2)
np.matmul(arr2, arr1) # (3, 3)
np.dot(arr1, arr2) # (2, 2)
arr1 @ arr2
```

## Reshaping

```python
arr = np.array([[-1, -2], [-3, -4], [-5, -6]]) # (3, 2)
arr.reshape([2, 3]) # เรียงลำดับตามเดิม ต้องมี element เท่ากัน 2*3 = 6
arr.reshape((2, 3)) # เรียงลำดับตามเดิม ต้องมี element เท่ากัน 2*3 = 6
arr.T # เปลี่ยนแถวเป็นคอลัมน์
```

## Array Level Operation

```python
# concatenation ก่อนต่อ
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
np.concatenate([arr1, arr2, arr1, arr2]) # list ต่อกันกี่ตัวก็ได้

arr3 = np.array([[1, 2, 3], [4, 5, 6]]) # (2, 3)
arr4 = np.array([[7, 8, 9], [10, 11, 12]]) # (2, 3)
result = np.concatenate([arr3, arr4])
result, result.shape # (4, 3) ต่อตัวที่ 1

# How about shape (2, 6)
result = np.concatenate([arr3, arr4], axis=0) # (2, 6) ต่อตัวที่ 2
result, result.shape

# max, min
l = [1,2,3,4]
max(l), min(l)

# max(arr3) - Error
np.max(arr3) # 6
np.max(arr3, axis=0) # array([4, 5, 6])
np.max(arr3, axis=1) # array([3, 6])

np.min(arr3) # 1
np.min(arr3, axis=0) # array([1, 2, 3])
np.min(arr3, axis=1) # array([1, 4)

# argmax, argmin - ตำแหน่งตัวที่มากที่สุด น้อยที่สุด
pred = np.array([0.1, 0.05, 0.85, 0.1, 0.04, 0.01, 0.05, 0.15, 0.009, 0.06])
np.argmax(pred) # 2
np.argmin(pred) # 8

np.argmax(arr3) # 5
np.argmax(arr3, axis=1) # array([2, 2]
np.argmax(arr4, axis=1) # array([2, 2]
# axis มีจำนวนไม่เกิน dimension
```
