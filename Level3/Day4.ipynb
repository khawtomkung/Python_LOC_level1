```python
import numpy as np

pred = np.array([0.5, 0.3, 0.1]) # 1-dimensional array of pred
target = np.array([0.2, 0.35, 0.1]) # 1-dimensional array of target
```

## Mean Square Error(MSE)
ค่าความคลาดเคลื่อน

np.sum((pred - target) ** 2) / len(pred) 
<!-- sum = zigma ผลรวม -->

```python
def MSE(pred: np.array, target: np.array):
    return np.sum((pred - target) ** 2) / len(pred) 
```

```python
MSE(pred, target)
```

## Mean Absolute Error (MAE)

```python
np.absolute(pred - target)
np.abs(pred - target)

def MSE(pred: np.array, target: np.array):
    return np.sum(np.abs(pred - target)) / len(pred) 
MSE(pred, target)
```
