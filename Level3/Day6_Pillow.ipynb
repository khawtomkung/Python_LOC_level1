```python
from PIL import Image
img = Image.open("myo.png")
```

```python
img.show() # img.
display(img)

img.size
img.info
img.width, img.height
# img = img.crop((400,100,900,400))
img.rotate(45, expand=True, fillcolor="white")
img.transpose(method=2)
img.convert('L')
```

```python
import numpy as np

img_arr = np.array(img)
img_inverted = np.array([255 - i for i in img_arr])
img_arr
img_inverted
Image.fromarray(img_inverted)
```

```python
img_arr_brighter = img_arr + 10
Image.fromarray(img_arr_brighter)
```

```python
from PIL import ImageEnhance
enhancer = ImageEnhance.Brightness(img)
brighter_img = enhancer.enhance(0.5)
brighter_img
# brighter_img.save("myo2.png")
```
