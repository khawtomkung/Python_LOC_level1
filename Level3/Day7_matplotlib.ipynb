```python
import plotly.express as px
import pandas as pd

df = pd.read_csv('Iris.csv')
df.head()
```

```python
fig = px.scatter(x=[1,3,7], y=[2,5,19])
fig.show()
```

```python
fig = px.scatter(data_frame=df, x="SepalWidthCm", y="PetalWidthCm", width=800, height=800, color="Species", size="PetalLengthCm")
fig.show()
```

```python
fig = px.scatter(data_frame=df, x="SepalWidthCm", y="PetalWidthCm", height=800, color="PetalLengthCm", symbol="Species", size="SepalLengthCm")
fig.show()
```
