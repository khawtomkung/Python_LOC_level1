```python
import plotly.express as px
import pandas as pd
```

```python
df = px.data.gapminder().query("country=='Canada'")
df
```

```python
fig = px.line(df, x="year", y="lifeExp")
fig.show()
```

```python
df = px.data.gapminder().query("continent=='Oceania'")
df
```

```python
fig = px.line(df, x="year", y="lifeExp", color="country", color_discrete_sequence=["pink", "blue"])
fig = px.line(df, 
              x="year", 
              y="lifeExp", 
              color="country", 
              color_discrete_map={"Australia": "#FE41D3", "New Zealand": "#432D4F"},
              markers=True,
              text="lifeExp",
              line_dash="country",
              title="Life Expectency of Oceania people",
              width=800
            )
fig.show()
```

## BAR CHART

```python
data_thai = px.data.gapminder().query("country=='Thailand'")
data_thai
```

```python
fig = px.bar(data_thai, x="year", y="pop", color="lifeExp", labels={"pop": "Population", "lifeExp": "Life Expectency"})
fig.show()
```

```python
data_tha_vie = px.data.gapminder().query("country=='Thailand' or country == 'Vietnam'")
data_tha_vie
```

```python
fig = px.bar(data_tha_vie, x="year", y="pop", color="country", barmode="group", text="gdpPercap", text_auto=".3s")
fig.update_traces(textfont_size=12, textangle=0, textposition="outside")
fig.show()
```

## PIE CHARTS

```python
data_asia = px.data.gapminder().query("continent == 'Asia' and year == 2007")
data_asia.loc[data_asia["pop"] < 2 * 10 ** 7, 'country'] = 'Other countries'
```

```python
fig = px.bar(data_asia, x="year", y="pop", color="country", barmode="group", text="gdpPercap", text_auto=".3s")
fig.update_traces(textfont_size=12, textangle=0, textposition="outside")
fig.show()
```

```python
fig = px.pie(data_asia, values="pop", names="country", color_discrete_sequence=px.colors.sequential.RdBu)
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```
